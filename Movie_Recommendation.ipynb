This cell imports the necessary libraries from NLTK (`nltk`) for natural language processing tasks: `stopwords` for removing common words and `WordNetLemmatizer` for reducing words to their base form. The `re` module is imported for regular expression operations.

This cell downloads the necessary NLTK data, specifically 'wordnet' for lemmatization and 'stopwords' for removing common words from text.

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('wordnet')
nltk.download('stopwords')
```

This cell initializes the `stopwords` set for English and a `WordNetLemmatizer` object, which will be used in text preprocessing.

```python
#stopwords
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

This cell defines a `preprocess_text` function that takes text as input, converts it to lowercase, removes punctuation, eliminates stopwords, and then lemmatizes the words to their base form. This prepares the text for feature extraction.

```python
def preprocess_text(text):
    #lowercase
    text = str(text).lower()
    #removed the punctuations
    text = re.sub(r'[^a-zA-Z\s]','', text)

    words = text.split()
    #remove stopwords
    words = [word for word in words if word not in stop_words]

    #lemmatize
    words = [lemmatizer.lemmatize(word) for word in words]

    return " ".join(words)
```

This cell applies the `preprocess_text` function to the 'tags' column of the DataFrame, cleaning and standardizing the text content for each movie.

```python
df['tags']= df['tags'].apply(preprocess_text)
```

This cell displays the preprocessed 'tags' for the movie at index 1, showcasing the result of the text cleaning steps.

```python
df['tags'][1]
```

This cell resets the DataFrame index after any potential row removals or modifications, ensuring a clean sequential index.

This cell creates a pandas Series called `indices` where movie titles are mapped to their corresponding DataFrame indices. This will be used to quickly retrieve a movie's index when searching for recommendations.

```python
indices = pd.Series(df.index,index=df['title']).drop_duplicates()
indices
```

This cell imports the `TfidfVectorizer` from `sklearn.feature_extraction.text`, which is used to convert a collection of raw documents to a matrix of TF-IDF features.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

This cell initializes the `TfidfVectorizer` with specific parameters: `max_features` limits the number of features, `ngram_range` considers single words and two-word phrases, and `stop_words='english'` removes common English stopwords during vectorization.

```python
tfidf = TfidfVectorizer(max_features=50000,ngram_range=(1,2),stop_words='english')
```

This cell fits the `TfidfVectorizer` to the preprocessed 'tags' column and transforms the text data into a TF-IDF matrix. Each row in this matrix represents a movie, and each column represents a TF-IDF score for a particular word or n-gram.

```python
tfidf_matrix = tfidf.fit_transform(df['tags'])
```

This cell displays the shape of the `tfidf_matrix`, showing the number of movies (rows) and the number of unique TF-IDF features (columns) extracted.

```python
tfidf_matrix
```

This cell imports `cosine_similarity` from `sklearn.metrics.pairwise`, a function used to calculate the cosine similarity between vectors. This metric will be used to determine the similarity between movies based on their TF-IDF representations.

```python
from sklearn.metrics.pairwise import cosine_similarity
```

This cell defines the `recommend` function, which takes a movie title and an optional number `n` (defaulting to 10) as input. It finds the movie's index, calculates its cosine similarity with all other movies, sorts them by similarity, and returns the top `n` similar movie titles.

```python
def recommend(title , n =10):
  if title not in indices:
    return ["Movie not found"]

  idx = indices[title]
  sim_score = cosine_similarity(tfidf_matrix[idx], tfidf_matrix).flatten()
  similar_idx = sim_score.argsort()[::-1][1:n+1]
  return df['title'].iloc[similar_idx]
```

This cell demonstrates the usage of the `recommend` function by asking for 5 recommendations similar to 'Avatar'.

```python
recommend('Avatar',5)
```
